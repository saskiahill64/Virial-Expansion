### Plotting alpha mass fraction for varying proton fractions

This code uses the generalised code from General_Fig_4 in order to plot x_alpha as a function of baryon density for varying proton fractions. This is done using the full model (including b_alpha coefficient).

```python
# Imports

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
```

```python
# Define constants

T = 4.0 # Temp [MeV]
m_p = 938.272 # mass of proton [MeV/c^2]
lamda = 197.326 * (2 * np.pi/(m_p * T))**(1/2) # nucleon thermal wavelength [fm^-1] assuming m_p=m_n
lamda_alpha = lamda / 2 # alpha particle thermal wavelength [fm^-1]
E_alpha = 28.3 # alpha particle binding energy [MeV]
Y_p_5 = 0.5 # proton fraction
Y_p_4 = 0.4
Y_p_3 = 0.3
Y_p_2 = 0.2
Y_p_1 = 0.1

# virial coefficients for T = 4 MeV
b_n = 0.307
b_pn = 3.19
b_nuc = 3.50
b_alpha_n = 2.73
b_alpha = 7.44
```

```python

# Calculate fugacities

#------------------------
# Y_p = 0.5
#------------------------

z_n_5 = np.linspace(1e-8, 0.05, 100) # range of neutron fugacities starting at small positive value

# with b_alpha solve quartic

z_p_5 = []

for zn in z_n_5:
    a = ((32 * Y_p_5 - 16)/(1 - Y_p_5)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p_5 - 16)/(1 - Y_p_5) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p_5)/(1 - Y_p_5)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p_5 - 8)/(1 - Y_p_5)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p_5)/(1 - Y_p_5)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p_5/(1 - Y_p_5)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e

    z_p = brentq(g, 0.0, 0.05)
    z_p_5.append(z_p)
z_p_5 = np.array(z_p_5)


z_alpha_5 = z_n_5**2 * z_p_5**2 * np.exp(E_alpha/T) # alpha particle fugacity

#------------------------
# Y_p = 0.4
#------------------------

z_n_4 = np.linspace(1e-8, 0.08, 100) # range of neutron fugacities starting at small positive value

# with b_alpha solve quartic

z_p_4 = []

for zn in z_n_4:
    a = ((32 * Y_p_4 - 16)/(1 - Y_p_4)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p_4 - 16)/(1 - Y_p_4) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p_4)/(1 - Y_p_4)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p_4 - 8)/(1 - Y_p_4)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p_4)/(1 - Y_p_4)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p_4/(1 - Y_p_4)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e

    z_p = brentq(g, 0.0, 0.05)
    z_p_4.append(z_p)
z_p_4 = np.array(z_p_4)


z_alpha_4 = z_n_4**2 * z_p_4**2 * np.exp(E_alpha/T) # alpha particle fugacity

#------------------------
# Y_p = 0.3
#------------------------

z_n_3 = np.linspace(1e-8, 0.1, 100) # range of neutron fugacities starting at small positive value

# with b_alpha solve quartic

z_p_3 = []

for zn in z_n_3:
    a = ((32 * Y_p_3 - 16)/(1 - Y_p_3)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p_3 - 16)/(1 - Y_p_3) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p_3)/(1 - Y_p_3)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p_3 - 8)/(1 - Y_p_3)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p_3)/(1 - Y_p_3)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p_3/(1 - Y_p_3)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e

    z_p = brentq(g, 0.0, 0.05)
    z_p_3.append(z_p)
z_p_3 = np.array(z_p_3)


z_alpha_3 = z_n_3**2 * z_p_3**2 * np.exp(E_alpha/T) # alpha particle fugacity

#------------------------
# Y_p = 0.2
#------------------------

z_n_2 = np.linspace(1e-8, 0.15, 100) # range of neutron fugacities starting at small positive value

# with b_alpha solve quartic

z_p_2 = []

for zn in z_n_2:
    a = ((32 * Y_p_2 - 16)/(1 - Y_p_2)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p_2 - 16)/(1 - Y_p_2) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p_2)/(1 - Y_p_2)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p_2 - 8)/(1 - Y_p_2)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p_2)/(1 - Y_p_2)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p_2/(1 - Y_p_2)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e

    z_p = brentq(g, 0.0, 0.05)
    z_p_2.append(z_p)
z_p_2 = np.array(z_p_2)


z_alpha_2 = z_n_2**2 * z_p_2**2 * np.exp(E_alpha/T) # alpha particle fugacity

#------------------------
# Y_p = 0.1
#------------------------

z_n_1 = np.linspace(1e-8, 0.2, 100) # range of neutron fugacities starting at small positive value

# with b_alpha solve quartic

z_p_1 = []

for zn in z_n_1:
    a = ((32 * Y_p_1 - 16)/(1 - Y_p_1)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p_1 - 16)/(1 - Y_p_1) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p_1)/(1 - Y_p_1)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p_1 - 8)/(1 - Y_p_1)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p_1)/(1 - Y_p_1)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p_1/(1 - Y_p_1)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e

    z_p = brentq(g, 0.0, 0.05)
    z_p_1.append(z_p)
z_p_1 = np.array(z_p_1)


z_alpha_1 = z_n_1**2 * z_p_1**2 * np.exp(E_alpha/T) # alpha particle fugacity
```

```python
# Calculate number densities

#------------------------------
# Y_p = 0.1
#------------------------------
n_n_1 = (2/lamda**3) * z_n_1 # neutron number density
n_p_1 = (2/lamda**3) * z_p_1 # proton number density
n_alpha_1 = (1/lamda_alpha**3) * z_alpha_1 # alpha particle number density
n_b_1 = (n_n_1 + n_p_1 + 4 * n_alpha_1) # baryon number density [fm^-3]

#------------------------------
# Y_p = 0.2
#------------------------------
n_n_2 = (2/lamda**3) * (z_n_2 + 2 * z_n_2**2 * b_n + 2 * z_n_2 * z_p_2 * b_pn)
n_p_2 = (2/lamda**3) * (z_p_2 + 2 * z_p_2**2 * b_n + 2 * z_n_2 * z_p_2 * b_pn)
n_alpha_2 = (1/lamda_alpha**3) * z_alpha_2
n_b_2 = (n_n_2 + n_p_2 + 4 * n_alpha_2)

#-----------------------------
# Y_p = 0.3
#-----------------------------
n_n_3 = (2/lamda**3) * (z_n_3 + 2 * z_n_3**2 * b_n + 2 * z_n_3 * z_p_3 * b_pn + 8 * z_n_3 * z_alpha_3 * b_alpha_n)
n_p_3 = (2/lamda**3) * (z_p_3 + 2 * z_p_3**2 * b_n + 2 * z_n_3 * z_p_3 * b_pn + 8 * z_p_3 * z_alpha_3 * b_alpha_n)
n_alpha_3 = (1/lamda_alpha**3) * (z_alpha_3 + 2 * z_alpha_3 * (z_n_3 + z_p_3) * b_alpha_n)
n_b_3 = (n_n_3 + n_p_3 + 4 * n_alpha_3)

#----------------------------
# Y_p = 0.4
#----------------------------
n_n_4 = (2/lamda**3) * (z_n_4 + 2 * z_n_4**2 * b_n + 2 * z_n_4 * z_p_4 * b_pn + 8 * z_n_4 * z_alpha_4 * b_alpha_n)
n_p_4 = (2/lamda**3) * (z_p_4 + 2 * z_p_4**2 * b_n + 2 * z_n_4 * z_p_4 * b_pn + 8 * z_p_4 * z_alpha_4 * b_alpha_n)
n_alpha_4 = (1/lamda_alpha**3) * (z_alpha_4 + 2 * z_alpha_4 * (z_n_4 + z_p_4) * b_alpha_n + 2 * z_alpha_4**2 * b_alpha)
n_b_4 = (n_n_4 + n_p_4 + 4 * n_alpha_4)

#----------------------------
# Y_p = 0.5
#----------------------------
n_n_5 = (2/lamda**3) * (z_n_5 + 2 * z_n_5**2 * b_n + 2 * z_n_5 * z_p_5 * b_pn + 8 * z_n_5 * z_alpha_5 * b_alpha_n)
n_p_5 = (2/lamda**3) * (z_p_5 + 2 * z_p_5**2 * b_n + 2 * z_n_5 * z_p_5 * b_pn + 8 * z_p_5 * z_alpha_5 * b_alpha_n)
n_alpha_5 = (1/lamda_alpha**3) * (z_alpha_5 + 2 * z_alpha_5 * (z_n_5 + z_p_5) * b_alpha_n + 2 * z_alpha_5**2 * b_alpha)
n_b_5 = (n_n_5 + n_p_5 + 4 * n_alpha_5)
```

```python
# Calculate alpha mass fractions


x_alpha_1 = 4 * n_alpha_1 / n_b_1 # alpha mass fraction
x_alpha_2 = 4 * n_alpha_2 / n_b_2 # alpha mass fraction
x_alpha_3 = 4 * n_alpha_3 / n_b_3 # alpha mass fraction
x_alpha_4 = 4 * n_alpha_4 / n_b_4 # alpha mass fraction
x_alpha_5 = 4 * n_alpha_5 / n_b_5 # alpha mass fraction
```

```python
# Plot results

plt.plot(n_b_1, x_alpha_1, label='Y_p = 0.1')
plt.plot(n_b_2, x_alpha_2, label='Y_p = 0.2')
plt.plot(n_b_3, x_alpha_3, label='Y_p = 0.3')
plt.plot(n_b_4, x_alpha_4, label='Y_p = 0.4')
plt.plot(n_b_5, x_alpha_5, label='Y_p = 0.5')
plt.xlabel('Baryon Density (n_b) [fm^-3]')
plt.ylabel('Alpha Mass Fraction (X_alpha)')
plt.title(f'Alpha Mass Fraction vs Baryon Density with varying Y_p')
plt.grid()
plt.legend()
plt.show()
```
