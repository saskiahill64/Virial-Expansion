# Panel plots of particle concentration against baryon density

```python
# Imports

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
```

```python
# Define constants

T = 4.0 # Temp [MeV]
m_p = 938.272 # mass of proton [MeV/c^2]
lamda = 197.326 * (2 * np.pi/(m_p * T))**(1/2) # nucleon thermal wavelength [fm^-1] assuming m_p=m_n
lamda_alpha = lamda / 2 # alpha particle thermal wavelength [fm^-1]
lamda_beta = lamda/np.sqrt(6) # beta particle thermal wavelength [fm^-1]
E_alpha = 28.3 # alpha particle binding energy [MeV]
E_beta = 29.27 # beta particle binding energy [MeV]
Y_p = 0.5 # proton fraction

# virial coefficients for T = 4 MeV
b_n = 0.307
b_pn = 3.19
b_nuc = 3.50
b_alpha_n = 2.73
b_alpha = 7.44
```

```python

# Calculate fugacities

z_n = np.linspace(1e-8, 0.05, 100) # range of neutron fugacities starting at small positive value

#------------------------------
# free n, p, alpha, beta gas
#------------------------------

# ideal quadratic coefficients

a_1 = (((16*Y_p - 8)/(1 - Y_p)) * z_n**2 * np.exp(E_alpha/T)) + (6*np.sqrt(6)*((3*Y_p - 1)/(1 - Y_p)) * z_n**4 * np.exp(E_beta/T))
b_1 = -1.0
c_1 = Y_p * z_n/(1 - Y_p)


z_p_1 = (2 * c_1) / (-b_1 + np.sqrt(b_1**2 - 4*a_1*c_1)) # proton fugacity calculated as roots of quadratic equation


z_alpha_1 = z_n**2 * z_p_1**2 * np.exp(E_alpha/T) # alpha particle fugacity

z_beta_1 = z_n**2 * z_alpha_1 * np.exp((E_beta - E_alpha)/T) # beta particle fugacity

#----------------------------
# full without He-6
#----------------------------

# solve quartic

z_p_2 = []

for zn in z_n:
    a = ((32 * Y_p - 16)/(1 - Y_p)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p - 16)/(1 - Y_p) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p)/(1 - Y_p)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p - 8)/(1 - Y_p)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) - 2 * b_n
    d = ((2 * Y_p)/(1 - Y_p)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p/(1 - Y_p)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e # quartic function in powers of z_p

    z_p = brentq(g, 0.0, 0.05)
    z_p_2.append(z_p)
z_p_2 = np.array(z_p_2)


z_alpha_2 = z_n**2 * z_p_2**2 * np.exp(E_alpha/T) # alpha particle fugacity

z_beta_2 = np.zeros_like(z_n)

#----------------------------
# full virial with He-6
#----------------------------

# solve quartic

z_p_3 = []

for zn in z_n:
    a = ((32 * Y_p - 16)/(1 - Y_p)) * b_alpha * zn**4 * np.exp(2 * E_alpha/T)
    b = (32 * Y_p - 16)/(1 - Y_p) * b_alpha_n * zn**2 * np.exp(E_alpha/T) - 8 * b_alpha_n * zn**2 * np.exp(E_alpha/T)
    c = ((8 * Y_p)/(1 - Y_p)) * b_alpha_n * zn**3 * np.exp(E_alpha/T) + ((16 * Y_p - 8)/(1 - Y_p)) * (zn**2 * np.exp(E_alpha/T) + 2 * zn**3 * np.exp(E_alpha/T) * b_alpha_n) + (6*np.sqrt(6)*((3*Y_p - 1)/(1 - Y_p)) * zn**4 * np.exp(E_beta/T)) - 2 * b_n
    d = ((2 * Y_p)/(1 - Y_p)) * zn * b_pn - 1 - 2 * (zn * b_pn)
    e = (Y_p/(1 - Y_p)) * (zn + 2 * zn**2 * b_n)

    def g(z_p):
        return a * z_p**4 + b * z_p**3 + c * z_p**2 + d * z_p + e # quartic function in powers of z_p

    z_p = brentq(g, 0.0, 0.1)
    z_p_3.append(z_p)
z_p_3 = np.array(z_p_3)


z_alpha_3 = z_n**2 * z_p_3**2 * np.exp(E_alpha/T) # alpha particle fugacity

z_beta_3 = z_n**2 * z_alpha_3 * np.exp((E_beta - E_alpha)/T) # beta particle fugacity
```

```python
# Calculate number densities

#------------------------------
# free n, p, alpha, beta gas
#------------------------------
n_n_1 = (2/lamda**3) * z_n # neutron number density
n_p_1 = (2/lamda**3) * z_p_1 # proton number density
n_alpha_1 = (1/lamda_alpha**3) * z_alpha_1 # alpha particle number density
n_beta_1 = (1/lamda_beta**3) * z_beta_1 # beta particle number density
n_b_1 = (n_n_1 + n_p_1 + 4 * n_alpha_1 + 6 * n_beta_1) # baryon number density [fm^-3]


#----------------------------
# full virial without He-6
#----------------------------
n_n_2 = (2/lamda**3) * (z_n + 2 * z_n**2 * b_n + 2 * z_n * z_p_2 * b_pn + 8 * z_n * z_alpha_2 * b_alpha_n)
n_p_2 = (2/lamda**3) * (z_p_2 + 2 * z_p_2**2 * b_n + 2 * z_n * z_p_2 * b_pn + 8 * z_p_2 * z_alpha_2 * b_alpha_n)
n_alpha_2 = (1/lamda_alpha**3) * (z_alpha_2 + 2 * z_alpha_2 * (z_n + z_p_2) * b_alpha_n + 2 * z_alpha_2**2 * b_alpha)
n_beta_2 = (1/lamda_beta**3) * z_beta_2 # array of zeros
n_b_2 = (n_n_2 + n_p_2 + 4 * n_alpha_2)

#----------------------------
# full virial with He-6
#----------------------------
n_n_3 = (2/lamda**3) * (z_n + 2 * z_n**2 * b_n + 2 * z_n * z_p_3 * b_pn + 8 * z_n * z_alpha_3 * b_alpha_n)
n_p_3 = (2/lamda**3) * (z_p_3 + 2 * z_p_3**2 * b_n + 2 * z_n * z_p_3 * b_pn + 8 * z_p_3 * z_alpha_3 * b_alpha_n)
n_alpha_3 = (1/lamda_alpha**3) * (z_alpha_3 + 2 * z_alpha_3 * (z_n + z_p_3) * b_alpha_n + 2 * z_alpha_3**2 * b_alpha)
n_beta_3 = (1/lamda_beta**3) * z_beta_3
n_b_3 = (n_n_3 + n_p_3 + 4 * n_alpha_3 + 6 * n_beta_3)
```

```python
# Calculate neutron mass fractions

x_n_1 = n_n_1 / n_b_1
x_n_2 = n_n_2 / n_b_2
x_n_3 = n_n_3 / n_b_3


# Calculate proton mass fractions

x_p_1 = n_p_1 / n_b_1
x_p_2 = n_p_2 / n_b_2
x_p_3 = n_p_3 / n_b_3


# Calculate alpha mass fractions


x_alpha_1 = 4 * n_alpha_1 / n_b_1 # alpha mass fraction
x_alpha_2 = 4 * n_alpha_2 / n_b_2 
x_alpha_3 = 4 * n_alpha_3 / n_b_3  


# Calculate He-6 mass fractions

x_he6_1 = 6 * n_beta_1 / n_b_1
x_he6_2 = 6 * n_beta_2 / n_b_2
x_he6_3 = 6 * n_beta_3 / n_b_3
```

```python
# Plot results for neutrons
plt.plot(n_b_1, x_n_1, label='free n, p, alpha, He-6 gas', linestyle='-.')
plt.plot(n_b_2, x_n_2, label='full virial without He-6', linestyle='--')
plt.plot(n_b_3, x_n_3, label='full virial with He-6', linestyle='-')
plt.xlabel('Baryon Density (n_b) [fm^-3]')
plt.ylabel('Neutron Mass Fraction (X_n)')
plt.title(f'n with Y_p = {Y_p}')
plt.grid()
plt.legend()
plt.show()


# Plot results for protons

plt.plot(n_b_1, x_p_1, label='free n, p, alpha, He-6 gas', linestyle='-.')
plt.plot(n_b_2, x_p_2, label='full virial without He-6', linestyle='--')
plt.plot(n_b_3, x_p_3, label='full virial with He-6', linestyle='-')
plt.xlabel('Baryon Density (n_b) [fm^-3]')
plt.ylabel('Proton Mass Fraction (X_alpha)')
plt.title(f'p with Y_p = {Y_p}')
plt.grid()
plt.legend()
plt.show()


# Plot results for alpha

plt.plot(n_b_1, x_alpha_1, label='free n, p, alpha, He-6 gas', linestyle='-.')
plt.plot(n_b_2, x_alpha_2, label='full virial without He-6', linestyle='--')
plt.plot(n_b_3, x_alpha_3, label='full virial with He-6', linestyle='-')
plt.xlabel('Baryon Density (n_b) [fm^-3]')
plt.ylabel('Alpha Mass Fraction (X_alpha)')
plt.title(f'alpha with Y_p = {Y_p}')
plt.grid()
plt.legend()
plt.show()


# Plot results for He-6

plt.plot(n_b_1, x_he6_1, label='free n, p, alpha, He-6 gas', linestyle='-.')
plt.plot(n_b_2, x_he6_2, label='full virial without He-6', linestyle='--')
plt.plot(n_b_3, x_he6_3, label='full virial with He-6', linestyle='-')
plt.xlabel('Baryon Density (n_b) [fm^-3]')
plt.ylabel('He-6 Mass Fraction (X_alpha)')
plt.title(f'He-6 with Y_p = {Y_p}')
plt.grid()
plt.legend()
plt.show()
```
